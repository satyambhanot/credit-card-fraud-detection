```python
# Notebook 01 – Exploratory Data Analysis (EDA)
# Purpose: Analyze class imbalance, scale relevant features,
# and examine correlations and value distributions
# Dataset: creditcard.csv

# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler

# Load the dataset
df = pd.read_csv('../data/raw/creditcard.csv')

# 1. Check Class Distribution (Legit vs Fraud)
plt.figure(figsize=(4, 3))
df['Class'].value_counts(normalize=True).plot(kind='bar', color=['skyblue', 'salmon'])
plt.title('Class Distribution')
plt.xticks([0, 1], ['Legit (0)', 'Fraud (1)'], rotation=0)
plt.ylabel('Proportion')
plt.tight_layout()
plt.show()

# This confirms significant class imbalance (fraud cases are a very small minority)

# 2. Scale the 'Amount' and 'Time' features
df['Amount_scaled'] = StandardScaler().fit_transform(df[['Amount']])
df['Time_scaled'] = StandardScaler().fit_transform(df[['Time']])

# All other features in the dataset are already standardized (PCA-transformed)

# 3. Generate correlation heatmap for all features
plt.figure(figsize=(12, 10))
sns.heatmap(df.corr(), cmap='coolwarm', center=0, linewidths=0.2, linecolor='gray')
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.show()

# This reveals correlation patterns and helps identify feature relationships or redundancy

# 4. Compare transaction amounts for Legit vs Fraud cases
plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x='Class', y='Amount_scaled', palette='pastel')
plt.title('Transaction Amount by Class')
plt.xticks([0, 1], ['Legit (0)', 'Fraud (1)'])
plt.ylabel('Scaled Amount')
plt.tight_layout()
plt.show()

# Useful for checking whether fraudulent transactions typically involve higher or lower amounts
```
